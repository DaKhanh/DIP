```python
import os
import faiss
import pickle
import pandas as pd
import json
from langchain.embeddings.openai import OpenAIEmbeddings
from langchain.vectorstores import FAISS
from langchain.docstore.in_memory import InMemoryDocstore
from langchain.schema import Document
from dotenv import load_dotenv

load_dotenv()

openai_api_key = os.getenv('OPENAI_API_KEY')
if not openai_api_key:
    raise ValueError("Missing OpenAI API key")

embeddings = OpenAIEmbeddings(openai_api_key=openai_api_key)

csv_path = 'D:\\dip_all\\app\\data\\data_cleaned.csv'

df = pd.read_csv(csv_path, index_col=0)

documents = []

for idx, row in df.iterrows():
    row_dict = row.to_dict()
    row_text = json.dumps(row_dict, ensure_ascii=False)
    doc = Document(page_content=row_text)
    documents.append(doc)

print("example:")
for i, doc in enumerate(documents[:3]):  
    print(f"\n--- doc {i+1} ---")
    print(doc.page_content)
    print(f"length: {len(doc.page_content)} characters")

import tiktoken

# 选择模型的编码方式，"text-embedding-ada-002" 适用于 text-embedding-3 模型系列
encoding = tiktoken.encoding_for_model("text-embedding-ada-002")

# 计算每个文本的 token 长度
print("Example with token counts:")
for i, doc in enumerate(documents[:3]):  
    tokens = encoding.encode(doc.page_content)  # 对文本进行编码
    token_count = len(tokens)  # 获取 token 数量
    print(f"\n--- doc {i+1} ---")
    print(doc.page_content)
    print(f"Character length: {len(doc.page_content)} characters")
    print(f"Token length: {token_count} tokens")
```
